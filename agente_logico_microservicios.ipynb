# Agente basado en conocimiento para la autorización de microservicios

## Inteligencia Artificial I

### Objetivo

Desarrollar un agente basado en conocimiento que permita evaluar las condiciones necesarias para autorizar el despliegue seguro de microservicios en una infraestructura Cloud.

El agente utilizará hechos, reglas lógicas y un motor de inferencia para determinar si se cumplen los requisitos de autorización.


```python
# Hechos iniciales del microservicio
hechos = {"P", "Q", "S"}

# Reglas que utiliza el agente para tomar decisiones
reglas = [
    ({"P", "Q"}, "R"),
    ({"R", "S"}, "T"),
    ({"T"}, "U")
]

# Mostrar la información registrada
print("Hechos iniciales:", hechos)
print("Reglas del agente:", reglas)
```

```python
# El agente revisa las reglas y obtiene nuevas conclusiones

def inferir(hechos, reglas):

    hubo_cambios = True

    while hubo_cambios:
        hubo_cambios = False

        for condiciones, conclusion in reglas:

            if condiciones.issubset(hechos):
                if conclusion not in hechos:

                    hechos.add(conclusion)
                    hubo_cambios = True

                    print("Nueva conclusión:", conclusion)

    return hechos
```

```python
# Aplicar las reglas a los hechos iniciales

print("INICIANDO EL RAZONAMIENTO DEL AGENTE")

hechos_finales = inferir(hechos, reglas)

print("\nBase de conocimiento actualizada:")
print(hechos_finales)
```

```python
# Consultar las decisiones tomadas por el agente

def consultar(hecho):

    if hecho in hechos_finales:
        print("ASK:", hecho, "-> VERDADERO")
    else:
        print("ASK:", hecho, "-> NO REGISTRADO")


print("CONSULTAS AL AGENTE")

consultar("R")
consultar("T")
consultar("U")
consultar("S")
consultar("V")
```

```python
# Registrar nueva información en la base de conocimiento

def tell(hecho):

    hechos_finales.add(hecho)

    print("TELL: Se registra el hecho", hecho)


print("REGISTRO DE NUEVA INFORMACION")

tell("V")

print("\nCONSULTA DESPUES DEL REGISTRO")

consultar("V")
```

## Arquitectura del agente basado en conocimiento (KB-Agent)

El agente desarrollado utiliza una base de conocimiento para almacenar hechos y reglas relacionadas con la autorización del despliegue de microservicios en una infraestructura Cloud.

Su funcionamiento se divide en tres procesos:

**1. Base de conocimiento:** almacena los hechos iniciales y las reglas que permiten evaluar las condiciones de autorización.

**2. Motor de inferencia:** utiliza encadenamiento hacia adelante para revisar las condiciones de las reglas y generar nuevas conclusiones cuando se cumplen los requisitos.

**3. Operaciones TELL y ASK:** TELL permite incorporar nuevos hechos a la base de conocimiento, mientras que ASK permite consultar si determinada información se encuentra registrada.

**Resultado:** el agente determinó que el microservicio cumple las condiciones para pasar a preproducción, obtener autorización de despliegue a producción y activar la autenticación multifactor (MFA) para los accesos administrativos.

## Punto A.: 
## Lógica proposicional

### Diccionario de proposiciones

las condiciones se establecen por las siguientes proposiciones:

- **P:** El microservicio supera las pruebas unitarias.
- **Q:** El microservicio supera la validación de seguridad estática.
- **R:** El microservicio queda habilitado para preproducción.
- **S:** El microservicio cuenta con la aprobación del líder técnico.
- **T:** Se autoriza el despliegue del microservicio a producción.
- **U:** Se activa la autenticación multifactor (MFA) para los accesos administrativos.

### las reglas de negocio


- **Regla 1:** (P ∧ Q) → R
- **Regla 2:** (R ∧ S) → T
- **Regla 3:** T → U

### Base de conocimiento (KB)

La base de conocimiento está formada por las tres reglas anteriores y los hechos iniciales:

- P: pruebas unitarias aprobadas.
- Q: validación de seguridad aprobada.
- S: aprobación del líder técnico.

Su representación conjunta es:

KB = {P, Q, S, (P ∧ Q) → R, (R ∧ S) → T, T → U}

### Demostración mediante Modus Ponens

**Primera deducción:**

Se conocen P y Q. Por tanto, se cumple la condición (P ∧ Q), y mediante la primera regla se obtiene R.

**Segunda deducción:**

Se conocen R y S. Por tanto, se cumple la condición (R ∧ S), y mediante la segunda regla se obtiene T.

**Tercera deducción:**

Como se obtuvo T, mediante la tercera regla se concluye U.

### Resultado

La base de conocimiento permite demostrar que el microservicio si cumple con las condiciones para ser habilitado en preproducción, recibir autorización de despliegue a producción y activar el MFA para los accesos administrativos.

## Enfoque 2: Lógica de primer orden (LPO)
### ¿Qué es la ontología?
Permite representar los elementos del problema mediante constantes, predicados y sus variables.

## 1. Identificación de elementos

Para representar el proceso de autorización se establece la siguiente ontología:

**Dominio:** microservicios de una infraestructura Cloud.

**Constante:**
- m1: representa un microservicio específico.

**Variable:**
- x: representa cualquier microservicio del sistema.

**Predicados:**
- Pruebas(x): el microservicio supera las pruebas unitarias.
- Seguridad(x): el microservicio supera la validación de seguridad estática.
- Preproduccion(x): el microservicio está habilitado para preproducción.
- Aprobacion(x): el microservicio cuenta con la aprobación del líder técnico.
- Produccion(x): el microservicio tiene autorización para desplegarse en producción.
- MFA(x): el microservicio tiene activada la autenticación multifactor para los accesos administrativos.

### 2. Representación de las reglas universales

Las reglas se expresan mediante cuantificadores universales:

**Regla 1:**

∀x [(Pruebas(x) ∧ Seguridad(x)) → Preproduccion(x)]

 Todo microservicio que supera las pruebas unitarias y la validación de seguridad será habilitado para preproducción.

**Regla 2:**

∀x [(Preproduccion(x) ∧ Aprobacion(x)) → Produccion(x)]

Todo microservicio se va habilitar para preproducción y aprobado por el líder técnico queda autorizado para producción.

**Regla 3:**

∀x [Produccion(x) → MFA(x)]

Estas expresiones indican que las condiciones se aplican a cualquier microservicio que pertenezca al dominio establecido.

### 3. Hechos iniciales

Se establecen los siguientes hechos:

- Pruebas(m1)
- Seguridad(m1)
- Aprobacion(m1)

Estos hechos indican que el microservicio m1 superó las pruebas unitarias, aprobó la validación de seguridad y cuenta con la aprobación del líder técnico.

### 4. Proceso de unificación

La unificación permite relacionar la variable x de las reglas con la constante m1.

La sustitución utilizada es:

θ = {x / m1}

Al aplicar esta sustitución a las reglas universales, se obtienen las siguientes expresiones:

- (Pruebas(m1) ∧ Seguridad(m1)) → Preproduccion(m1)
- (Preproduccion(m1) ∧ Aprobacion(m1)) → Produccion(m1)
- Produccion(m1) → MFA(m1)

### 5. Resultado de la inferencia

A partir de los hechos iniciales y aplicando las reglas mediante la sustitución θ, se obtiene:

1. Preproduccion(m1).
2. Produccion(m1).
3. MFA(m1).

la lógica de primer orden permite representar las reglas de autorización de manera general, utilizando variables y predicados, y aplicarlas a un microservicio específico mediante un proceso de unificación.

## Comparación de los tres enfoques lógicos

| Enfoque | Característica principal | Aplicación |
|---|---|---|
| Lógica proposicional | Utiliza proposiciones y conectores lógicos. | Representa las condiciones para autorizar el despliegue. |
| Lógica de primer orden | Utiliza variables, predicados y cuantificadores. | Permite aplicar reglas generales a diferentes microservicios. |
| Agente basado en conocimiento | Almacena hechos, aplica reglas y responde consultas. | Automatiza las decisiones mediante inferencias. |

### Conclusión

Los tres enfoques permiten representar el proceso de autorización de microservicios desde diferentes perspectivas. 

La implementación realizada en Python permitió comprobar que, cuando se cumplen las condiciones iniciales, el agente puede inferir la habilitación para preproducción, la autorización de producción y la activación de MFA.

CitaRegistrada(c,v,f,h) ∧ SolicitaCancelacion(c,v,f,h) → CancelacionAprobada(c,v,f,h).

**Regla 4. Aprobación de una reprogramación**

SolicitaReprogramacion(c,v,f,h) ∧ Disponibilidad(v,f,h) ∧ SinConflicto(v,f,h) → ReprogramacionAprobada(c,v,f,h).

**Regla 5. Una alternativa**

Si el cliente solicita reprogramar una cita, pero existe un conflicto en el horario solicitado, el agente debe buscar una alternativa.

SolicitaReprogramacion(c,v,f,h) ∧ Conflicto(v,f,h) → BuscarAlternativa(c,v,f,h).

# 3. Implementación del agente basado en conocimiento

En esta etapa se desarrollará un agente inteligente en Python que utilizará una base de conocimiento para analizar las solicitudes de citas veterinarias.

El agente estará compuesto por tres elementos principales:

- **Base de conocimiento:** almacena los hechos y las reglas del sistema.
- **Motor de inferencia:** evaluará las condiciones y obtendrá nuevas conclusiones.
- **Sistema de consulta:** permitirá verificar las decisiones tomadas por el agente.

La implementación se realizará mediante encadenamiento hacia adelante, es decir, el agente revisará los hechos disponibles y aplicará las reglas que se cumplan para generar nuevas conclusiones.

## 3.1. Preparación del entorno y estructura de la base de conocimiento

En esta etapa se utilizarán estructuras de Python para almacenar los hechos y las reglas que representan el funcionamiento del sistema veterinario.

La base de conocimiento se organizará mediante conjuntos y listas, lo que permitirá registrar información, consultar datos y aplicar las reglas establecidas para cada solicitud.

```python
# Base de conocimiento del agente veterinario

# Aquí se guardarán los datos que conoce el agente
hechos = set()

# En esta lista se almacenarán las reglas de decisión
reglas = []

print("Base de conocimiento preparada.")
print("Hechos registrados:", hechos)
print("Reglas definidas:", reglas)
```

```python
# Reglas que utilizará el agente para tomar decisiones

reglas = [
    {
        "nombre": "Aprobacion de reserva",
        "requiere": [
            "Cliente",
            "Veterinario",
            "SolicitaReserva",
            "Disponibilidad",
            "SinConflicto"
        ],
        "concluye": "ReservaAprobada"
    },
    {
        "nombre": "Deteccion de conflictos",
        "requiere": [
            "SolicitaReserva",
            "Conflicto"
        ],
        "concluye": "BuscarAlternativa"
    },
    {
        "nombre": "Aprobacion de cancelacion",
        "requiere": [
            "CitaRegistrada",
            "SolicitaCancelacion"
        ],
        "concluye": "CancelacionAprobada"
    },
    {
        "nombre": "Aprobacion de reprogramacion",
        "requiere": [
            "SolicitaReprogramacion",
            "Disponibilidad",
            "SinConflicto"
        ],
        "concluye": "ReprogramacionAprobada"
    },
    {
        "nombre": "Alternativa para reprogramacion",
        "requiere": [
            "SolicitaReprogramacion",
            "Conflicto"
        ],
        "concluye": "BuscarAlternativa"
    }
]

print("Reglas cargadas correctamente:")

for regla in reglas:
    print("-", regla["nombre"])
```

```python
# Motor de inferencia del agente veterinario
# Revisa los hechos y aplica las reglas que se cumplan.

def aplicar_reglas(hechos, reglas):

    # Copiamos los hechos para conservar los datos originales
    nuevos_hechos = set(hechos)

    # Guardamos las decisiones que va tomando el agente
    decisiones = []

    hubo_cambios = True

    # Se repite el proceso mientras aparezcan nuevas conclusiones
    while hubo_cambios:
        hubo_cambios = False

        for regla in reglas:

            condiciones = set(regla["requiere"])
            conclusion = regla["concluye"]

            # Verificamos si se cumplen todas las condiciones
            if condiciones.issubset(nuevos_hechos):

                # Evitamos registrar dos veces la misma conclusión
                if conclusion not in nuevos_hechos:

                    nuevos_hechos.add(conclusion)
                    decisiones.append(conclusion)
                    hubo_cambios = True

                    print("Regla aplicada:", regla["nombre"])
                    print("Decision generada:", conclusion)
                    print("--------------------------------")

    return nuevos_hechos, decisiones
```

```python
# Escenario 1: solicitud de una nueva reserva

hechos_escenario_1 = {
    "Cliente",
    "Veterinario",
    "SolicitaReserva",
    "Disponibilidad",
    "SinConflicto"
}

print("Evaluando solicitud de reserva...")
print("Hechos iniciales:", hechos_escenario_1)

resultado_1, decisiones_1 = aplicar_reglas(
    hechos_escenario_1,
    reglas
)

print("\nResultado del escenario 1:")

for decision in decisiones_1:
    print("-", decision)
```
